```python
import csv
#Name of the csv file
file_name = "expenses.csv"

# function to add a new expense
def add_expense():

   # Take expense details from the user
   date = input("Enter date: ")
   category = input("Enter category: ")


   # Take amount and handle invalid input
   try:
     amount = float(input("Enter amount: "))
   except ValueError:
     print("Please enter a valid amount.")
     return

   # Note is optional
   note = input("Enter note: ")
   # Open the CSV file in append mode
   with open(file_name, "a", newline="") as file:
     # Create a CSV writer
     writer = csv.writer(file)

     # Write the expense into the file
     writer.writerow([date, category, amount, note])
   print("Expense added successfully!")

# Function to view all expenses
def view_expenses():
   try:
     # Open the CSV file for reading
     with open(file_name, "r") as file:
     # Read the CSV file
     reader = csv.reader(file)
     # Variable to store total spending
     total = 0
     print("\nAll Expenses")

   # Read each expense one by one
   for row in reader:
     print("Date:", row[0])
     print("Category:", row[1])
     print("Amount:", row[2])
     print("Note:", row[3])
     print("--------------------")

     # Add the amount to the total
     total = total + float(row[2])

   # Display total spending
   print("Total Amount Spent:", total)

   # Handle the case when the file does not exist
   except FileNotFoundError:
     print("No expense file found.")

# Function to display category-wise spending
def category_summary():
   try:
     # Open the CSV file for reading
     with open(file_name, "r") as file:
       # Read the CSV file
       reader = csv.reader(file)
       # Variables to store category totals
       food = 0
       travel = 0
       shopping = 0
       other = 0
       # Read each expense
       for row in reader:
         category = row[1]
         amount = float(row[2])

         # Add amount according to category
         if category == "Food":
           food = food + amount
         elif category == "Travel":
           travel = travel + amount
         elif category == "Shopping":
           shopping = shopping + amount
         else:
           other = other + amount

     # Display the category-wise totals
     print("\nCategory-wise Summary")
     print("Food:", food)
     print("Travel:", travel)
     print("Shopping:", shopping)
     print("Other:", other)

   # Handle missing file
   except FileNotFoundError:
    print("No expense file found.")


# Main function
def main():
  # Keep showing the menu until the user chooses Exit
  while True:
    print("\nExpense Tracking System")
    print("1. Add Expense")
    print("2. View Expenses")
    print("3. Category-wise Summary")
    print("4. Exit")

    # Handle invalid menu input
try:
  choice = int(input("Enter your choice: "))
  # Add an expense
  if choice == 1:
    add_expense()

  # View all expenses
  elif choice == 2:
    view_expenses()

  # Show category-wise spending
  elif choice == 3:
    category_summary()
  # Exit the program
  elif choice == 4:
    print("Thank you for using Expense Tracker.")
    break
  # Handle numbers other than 1 to 4
  else:
    print("Invalid choice.")
 # Handle non-numeric input
 except ValueError:
 print("Please enter a number.")

#Start the program
main()
```
